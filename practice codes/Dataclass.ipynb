```python
@dataclass
class Point:
    x:int
    y:int
```

```python
p= Point(1,2)
```

```python
print(p)
```

```python
print(p == Point(1, 2))
```

```python
class Bad:
    items = []

a,b = Bad(),Bad()

a.items.append("x")
```

```python
print(b.items)
```

```python
# Above is Giving same (which is wrong) out put bec they are sharing class attributes.
```

```python
from dataclasses import field

@dataclass
class Good:
    items: list = field(default_factory=list)  # Generates a new list for each instance

a,b = Good(),Good()

a.items.append("x")
```

```python
print(b.items)
```
